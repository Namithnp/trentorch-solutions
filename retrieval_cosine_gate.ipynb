Story
A retrieval service ranks document embeddings against a query. One code path uses raw dot product while another uses cosine similarity, causing inconsistent ranking when vector norms differ. Compute cosine similarity for each document vector.

The Math
For query q and document vector d_i:

$$s_i = (q . d_i) / (||q||_2 * ||d_i||_2)$$
Copy
If either vector has zero norm, define its similarity as 0.

Input Format
query (length d), documents (n x d).

Output Format
n cosine similarities, 6 decimal places.

Constraints
1 <= n <= 10^4, 1 <= d <= 1000, component magnitudes at most 10^6.
Time limit: 2.0 seconds.
Example
Input

query = [1, 0], documents = [[3, 0], [1, 1]]
Copy
Output

1.000000 0.707107
Copy
Explanation: The first document points in the same direction as the query (similarity 1). The second forms a 45-degree angle with it.

```python
query = torch.tensor([1., 0., 1., 0.])

docs = torch.tensor([
    [ 1., 0.,  1., 0.],   # same as query
    [ 0., 1.,  0., 1.],   # orthogonal
    [ 2., 0.,  2., 0.],   # same direction, 2x longer
    [ 1., 1.,  0., 0.],   # partly aligned
    [-1., 0., -1., 0.],   # opposite
])

norm1 = np.linalg.norm(query)
norm2 = np.linalg.norm(docs)

print(norm1, norm2)
```

```python
import numpy as np


def retrieval_cosine_gate(query: np.ndarray, documents: np.ndarray) -> np.ndarray:
    """
    query: array of shape (d,).
    documents: array of shape (n, d).

    Returns an array of shape (n,): cosine similarity between query and
    each document row. If either the query or a document has zero norm,
    that document's similarity is 0.
    """
    # TODO: compute dot(query, doc) / (||query|| * ||doc||) per document,
    # guarding against zero norms.
    
    cos_sim = np.array([])
    for doc in documents:
        s_i = (np.dot(query, doc)) / (np.linalg.norm(query) * np.linalg.norm(doc)) if (np.linalg.norm(query) != 0 and np.linalg.norm(doc) != 0) else 0
        cos_sim = np.append(cos_sim, s_i)
    return cos_sim
```
